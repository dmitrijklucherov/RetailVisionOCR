```python

def filter_zerro(new):
    new_baza = []
    for i in new:
        id_transaction, id_customer, category, items, unit_price, quantity, total_spent, pay_method, location, tr_date, dis_applied = i
        if unit_price == 0.0 or quantity == 0.0 or total_spent == 0:
            continue
        else:
            new_baza.append(i)
    return new_baza
```

```python
import psycopg
def connect():
    connection = psycopg.connect(
        dbname = "postgres", user = "postgres", password = "password", host = "127.0.0.1", port = "5432"
    )
    cursor = connection.cursor()
    query = """select id_transaction, id_customer, category, items, unit_price, quantity, total_spent, pay_method, location, tr_date, dis_applied  from retail_store_sales;"""
    cursor.execute(query)
    import_query = cursor.fetchall()
    cursor.close()
    connection.close()
    return import_query
```

```python
raw_data = connect()
print(f"Успешно выгружено: {len(raw_data)} строк!")
```

```python
final_baza = filter_zerro(raw_data)
print(f"Анализ завершен! Итоговое количество строк: {len(final_baza)} строк")
print(final_baza[25])
```
